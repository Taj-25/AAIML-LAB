Implement a Matrix factorization on a suitable dataset.

```python
import pandas as pd
from sklearn.decomposition import NMF
import numpy as np

ratings = pd.read_csv('ratings.csv')  # Contains userId, movieId, rating, and timestamp
movies = pd.read_csv('movies.csv')  # Contains movieId, title, and genres

# Create ratings matrix => pivot table( reorganize data in a structured way)
ratings_matrix = ratings.pivot(index='userId', columns='movieId', values='rating')

# Fill NaN values with the mean of each column (movieId)
for column in ratings_matrix.columns:
    ratings_matrix[column] = ratings_matrix[column].fillna(ratings_matrix[column].mean())

# Define the NMF model 
model = NMF(n_components=25, init='random', random_state=0, max_iter=500, tol=1e-4)

# Factorize the matrix by applying the NMF
# W: (number of users, n_components)
# H: (n_components, number of movies)
W = model.fit_transform(ratings_matrix)  # User matrix
H = model.components_  # Movie matrix

# Save the matrices to CSV files
pd.DataFrame(W, index=ratings_matrix.index).to_csv('user_matrix.csv', index=True)
pd.DataFrame(H, columns=ratings_matrix.columns).to_csv('movie_matrix.csv', index=False)

# Optionally: Merge movie details with movie matrix for better analysis
movie_features = pd.DataFrame(H, columns=ratings_matrix.columns)
movie_features = movie_features.transpose().reset_index()
movie_features.columns = ['movieId'] + [f'Feature_{i}' for i in range(1, 26)]
merged_movies = pd.merge(movie_features, movies, on='movieId', how='inner')

# Save the merged movie details
merged_movies.to_csv('movie_features_with_details.csv', index=False)
print("Successfully created Factorized Matrices -> 1.UserMatrix 2. MovieMatrix 3.MovieFeatureMatrix")
```
